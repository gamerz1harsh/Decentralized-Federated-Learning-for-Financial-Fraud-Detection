```python
# ==========================================================
# Data Preprocessing for Credit Card Fraud Detection Dataset
# ==========================================================

# Import required libraries for data manipulation,
# visualization, preprocessing, and train-test splitting.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# ==========================================================
# Paths (dynamic - resolved relative to the project root)
# Place your raw dataset CSV inside  data/raw/  and update
# the filename below if your file has a different name.
# ==========================================================
from pathlib import Path

# Notebook lives in <project_root>/data/, so project root is one level up.
PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name.lower() == 'data':
    PROJECT_ROOT = PROJECT_ROOT.parent

RAW_PATH = PROJECT_ROOT / 'data' / 'raw' / 'creditcard.csv'

if not RAW_PATH.exists():
    raise FileNotFoundError(
        f'Raw dataset not found at: {RAW_PATH}\n'
        'Please download the credit card fraud dataset (creditcard.csv) '
        'and place it in the data/raw/ folder of this project.'
    )

# Load the raw dataset from disk.
df = pd.read_csv(RAW_PATH)
print(f'Loaded dataset from: {RAW_PATH}')
```

```python
# Remove duplicate rows from the dataset.
print("Shape Before:", df.shape)

df = df.drop_duplicates()

print("Shape After :", df.shape)
```

```python
# Confirm the dataset no longer contains duplicates.
print("Duplicates:", df.duplicated().sum())
```

```python
# Separate features and target variables.
X = df.drop(columns=["Class"])
y = df["Class"]
```

```python
# Check the shapes of the feature and target arrays.
print(X.shape)
print(y.shape)
```

```python
# Split the dataset into training and testing sets.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
# Inspect the distribution of labels in the train and test sets.
print(X_train.shape)
print(X_test.shape)

print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

```python
# Standardize the Time and Amount features using the training data.
scaler = StandardScaler()

X_train[["Time", "Amount"]] = scaler.fit_transform(
    X_train[["Time", "Amount"]]
)

X_test[["Time", "Amount"]] = scaler.transform(
    X_test[["Time", "Amount"]]
)
```

```python
# Review the scaled feature statistics.
print(X_train[["Time", "Amount"]].describe())
```

```python
# Create DataFrames for the processed training and testing sets.
train_df = X_train.copy()
train_df["Class"] = y_train.values

test_df = X_test.copy()
test_df["Class"] = y_test.values
```

```python
# Save the processed training and testing data to disk.
from pathlib import Path

processed_dir = Path("../data/processed")
processed_dir.mkdir(exist_ok=True)

train_df.to_csv(processed_dir / "train.csv", index=False)
test_df.to_csv(processed_dir / "test.csv", index=False)
```
